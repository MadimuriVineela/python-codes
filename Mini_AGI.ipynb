```python
import json
import os

class MiniAGI:
    def __init__(self, memory_file="memory.json"):
        self.memory_file = memory_file
        self.memory = self.load_memory()

    def load_memory(self):
        if os.path.exists(self.memory_file):
            with open(self.memory_file, "r") as f:
                return json.load(f)
        return {}

    def save_memory(self):
        with open(self.memory_file, "w") as f:
            json.dump(self.memory, f, indent=4)

    def learn(self, key, value):
        self.memory[key] = value
        self.save_memory()

    def reason(self, input_text):
        input_text = input_text.lower()

        if "my name is" in input_text:
            name = input_text.split("my name is")[-1].strip()
            self.learn("user_name", name)
            return f"Nice to meet you, {name}!"

        if "what is my name" in input_text:
            return self.memory.get("user_name", "I don't know your name yet.")

        if "remember" in input_text:
            data = input_text.replace("remember", "").strip()
            self.learn("fact", data)
            return "I have remembered that."

        if "what did you remember" in input_text:
            return self.memory.get("fact", "I have not learned anything yet.")

        if "add" in input_text:
            nums = [int(s) for s in input_text.split() if s.isdigit()]
            return f"Sum is {sum(nums)}"

        return "I am learning. Please teach me more."

# Run Mini AGI
agi = MiniAGI()
print("Mini AGI is running (type 'exit' to stop)\n")

while True:
    user_input = input("You: ")
    if user_input.lower() == "exit":
        break
    response = agi.reason(user_input)
    print("AGI:", response)
```
